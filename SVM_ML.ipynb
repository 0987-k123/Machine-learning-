```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
plt.style.use('ggplot')
```

```python
from sklearn.datasets import load_breast_cancer

bcancer = load_breast_cancer()

print("Features:")
print(bcancer.feature_names)

print("Target:")
print(bcancer.target)
```

```python
bcancer.data.shape
```

```python
X__train, X__test, y__train, y__test = train_test_split(bcancer.data, bcancer.target, test_size=0.3, random_state=109)
```

```python
from sklearn import svm
clf = svm.SVC(kernel='linear')
clf.fit(X__train, y__train)

y_pred = clf.predict(X__test)
print("modelBuilding Completed")
```

```python
from sklearn import metrics
print("Accuracy:", metrics.accuracy_score(y__test, y_pred))
print("Precision:", metrics.precision_score(y__test, y_pred))
print("Recall:", metrics.recall_score(y__test, y_pred))
print("F1 Score:", metrics.f1_score(y__test, y_pred))
```
